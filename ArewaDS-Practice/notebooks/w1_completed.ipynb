## Exercise 4.1: Temperature Converter

Convert Celsius to Fahrenheit using:

`fahrenheit = (celsius * 9/5) + 32`


```python
# Exercise 4.1
celsius = float(input("Enter a temperature in Celsius: "))
fahrenheit = (celsius * 9 / 5) + 32

print("Temperature in Fahrenheit:", fahrenheit)
```

## Exercise 4.2: Simple Calculator

Ask for two numbers and calculate:
- sum
- difference
- product
- quotient


```python
# Exercise 4.2
first_number = float(input("Enter the first number: "))
second_number = float(input("Enter the second number: "))

sum_result = first_number + second_number
difference = first_number - second_number
product = first_number * second_number

print("Sum:", sum_result)
print("Difference:", difference)
print("Product:", product)

if second_number != 0:
    quotient = first_number / second_number
    print("Quotient:", quotient)
else:
    print("Quotient: undefined because division by zero is not allowed.")
```

## Challenge 1: Area Calculator

Ask for the rectangle length and width, then calculate and print:
- area
- perimeter


```python
# Challenge 1
length = float(input("Enter the rectangle length: "))
width = float(input("Enter the rectangle width: "))

area = length * width
perimeter = 2 * length + 2 * width

print("Area:", area)
print("Perimeter:", perimeter)
```

## Challenge 2: Shopping Bill Calculator

Ask for the prices of three items, then calculate:
- subtotal
- 6% sales tax
- final total


```python
# Challenge 2
item1 = float(input("Enter the price of item 1: "))
item2 = float(input("Enter the price of item 2: "))
item3 = float(input("Enter the price of item 3: "))

subtotal = item1 + item2 + item3
sales_tax = subtotal * 0.06
final_total = subtotal + sales_tax

print("Subtotal:", subtotal)
print("Sales tax:", sales_tax)
print("Final total:", final_total)
```
